```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("iabhishekofficial/mobile-price-classification")
import pandas as pd

df = pd.read_csv(path + "/train.csv")
```

```python
df_test = pd.read_csv(path + "/test.csv")
df_test.head()
```

```python
X = df.drop(columns=["price_range"])
y = df["price_range"]

cat_cols = ['blue', 'dual_sim', 'four_g', 'three_g', 'touch_screen', 'wifi']
num_cols = [col for col in X.columns if col not in cat_cols]


```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler

preprocessor = ColumnTransformer(
    [
        ("num", StandardScaler(), num_cols),
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from lightgbm import LGBMClassifier

pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LGBMClassifier(n_estimators=100, num_leaves=31, learning_rate=0.05, max_depth=5, min_child_samples=20, random_state=42))
])

giper_params= {
    "model__n_estimators": [50, 100, 200, 300],
    "model__num_leaves": [31, 63, 127],
    "model__learning_rate": [0.01, 0.05, 0.1],
    "model__max_depth": [3, 5, 8],
    "model__min_child_samples": [20, 50, 100]
}

from sklearn.model_selection import GridSearchCV

GridSearch = GridSearchCV(pipeline, giper_params, cv=5, scoring="accuracy", n_jobs=-1, verbose=2)
```

```python
GridSearch.fit(X, y)

print("Best parameters:", GridSearch.best_params_)
print("Best score:", GridSearch.best_score_)
```
