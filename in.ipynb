```python
import pandas as pd

df = pd.read_csv('loan_status.csv')
df.head()
```

```python
df.columns

# (['Loan_ID', 'Jinsi', 'Uylangan', 'Qarindoshlar', 'Ma'lumoti', 'O'zini o'zi ish bilan ta'minlagan', 'Ariza beruvchining daromadi', 'Hamkorlikdagi ariza beruvchining daromadi', 'LoanAmount', 'Loan_Amount_Term', 'Kredit_Tarixi', 'Mulk_Maydoni', 'Loan_Holati'],
# dtype='str')
```

```python
df.columns = ['Loan_ID', 'Jinsi', 'Uylangan', 'Qarindoshlari soni', "Ma'lumoti", "O'zini o'zi ish bilan ta'minlagan", 'Ariza beruvchining daromadi', 'Hamkorlikdagi ariza beruvchining daromadi', 'Kredit miqdori', 'Kredit miqdorining muddati', 'Kredit Tarixi', 'Mulk Maydoni', 'Loan_Holati']
```

```python
df.head()
```

```python
df.drop(['Loan_ID'], axis=1, inplace=True)
```

```python
df.info()
```

```python
for col in df.columns:
    print(f"{col}: {df[col].isnull().sum()}")
```

```python
df['Jinsi'] = df['Jinsi'].fillna(df['Jinsi'].mode()[0], inplace=True)
```

```python
import numpy as np

shart = df['Qarindoshlari soni'] == 0

# np.where natijasini Series'ga o'giramiz va indeksni df bilan bir xil qilamiz
toldiruvchi_qiymatlar = pd.Series(np.where(shart, 'No', 'Yes'), index=df.index)

df['Uylangan'] = df['Uylangan'].fillna(toldiruvchi_qiymatlar)
```

```python
df['Qarindoshlari soni'] = df['Qarindoshlari soni'].fillna(df['Qarindoshlari soni'].mode()[0], inplace=True)
```

```python
df["O'zini o'zi ish bilan ta'minlagan"] = df["O'zini o'zi ish bilan ta'minlagan"].fillna(df["O'zini o'zi ish bilan ta'minlagan"].mode()[0], inplace=True)
```

```python
df['Kredit miqdori'] = df['Kredit miqdori'].fillna(df['Kredit miqdori'].median(), inplace=True)
```

```python
df['Kredit miqdorining muddati'] = df['Kredit miqdorining muddati'].fillna(df['Kredit miqdorining muddati'].median(), inplace=True)
df['Kredit Tarixi'] = df['Kredit Tarixi'].fillna(df['Kredit Tarixi'].mode()[0], inplace=True)
```

```python
df.info()
```

```python
df.describe()
```

```python
from sklearn.model_selection import train_test_split

X = df.drop('Loan_Holati', axis=1)
Y = df['Loan_Holati']

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

num_cols = ['Ariza beruvchining daromadi', 'Hamkorlikdagi ariza beruvchining daromadi', 'Kredit miqdori', 'Kredit miqdorining muddati']
cat_cols = [col for col in X.columns if col not in num_cols]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

Transformer = ColumnTransformer(transformers=[
    ('scaler', StandardScaler(), num_cols),
    ('encoder', OneHotEncoder(), cat_cols)
])
   
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
Y_train = le.fit_transform(Y_train)
Y_test = le.transform(Y_test)
```

```python
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline

svc_best_params = {
    'classifier__C': [1.0, 10.0, 100.0],
    'classifier__gamma': [0.01, 0.1, 1.0],
    'classifier__kernel': ['linear', 'rbf', 'poly']
}

svc_pipeline = Pipeline(steps=[
    ('preprocessor', Transformer),
    ('classifier', SVC())
])
svc_grid_search = GridSearchCV(svc_pipeline, svc_best_params, cv=5, scoring='accuracy', n_jobs=-1)

svc_grid_search.fit(X_train, Y_train)

print("Best parameters for SVC:", svc_grid_search.best_params_)
print("Best cross-validation accuracy for SVC:", svc_grid_search.best_score_)

```

```python
from sklearn.neighbors import KNeighborsClassifier

knn_best_params = {
    'classifier__n_neighbors': [3, 5, 7, 9, 11],
    'classifier__weights': ['uniform', 'distance'],
    'classifier__metric': ['euclidean', 'manhattan', 'minkowski']
}


knn_pipeline = Pipeline(steps=[
    ('preprocessor', Transformer),
    ('classifier', KNeighborsClassifier())
])
knn_grid_search = GridSearchCV(knn_pipeline, knn_best_params, cv=5, scoring='accuracy', n_jobs=-1)
knn_grid_search.fit(X_train, Y_train)

print("Best parameters for KNN:", knn_grid_search.best_params_)
print("Best cross-validation accuracy for KNN:", knn_grid_search.best_score_)
```

```python
from scipy.stats import loguniform, randint
from catboost import CatBoostClassifier
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report

# target: 'N','Y' -> 0,1
le = LabelEncoder()
Y_train_enc = le.fit_transform(Y_train)
Y_test_enc = le.transform(Y_test)

pipeline = Pipeline(steps=[
    ('preprocessor', Transformer),
    ('classifier', CatBoostClassifier(verbose=0, random_state=42))
])

param_distributions = {
    'classifier__iterations': randint(200, 1500),
    'classifier__learning_rate': loguniform(0.01, 0.3),
    'classifier__depth': randint(4, 10),
    'classifier__l2_leaf_reg': loguniform(1, 20),
    'classifier__random_strength': loguniform(0.1, 10),
    'classifier__bagging_temperature': [0, 0.5, 1, 3],
}

cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

Catboost_search = RandomizedSearchCV(
    pipeline,
    param_distributions=param_distributions,
    n_iter=20,
    cv=cv,
    scoring='accuracy',
    n_jobs=1,          # CatBoost o'zi ko'p yadroli ishlaydi
    random_state=42,
    verbose=2,
    refit=True,
)

Catboost_search.fit(X_train, Y_train_enc)

print("Best parameters:", Catboost_search.best_params_)
print("Best CV accuracy:", round(Catboost_search.best_score_, 4))
```

```python
from scipy.stats import loguniform, randint, uniform
from xgboost import XGBClassifier
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import LabelEncoder

# target: 'N','Y' -> 0,1 (XGBoost faqat 0/1 qabul qiladi)
le = LabelEncoder()
Y_train_enc = le.fit_transform(Y_train)
Y_test_enc = le.transform(Y_test)

pipeline = Pipeline(steps=[
    ('preprocessor', Transformer),
    ('classifier', XGBClassifier(
        eval_metric='logloss',
        random_state=42,
        n_jobs=1,
    ))
])

param_distributions = {
    'classifier__n_estimators': randint(200, 1500),
    'classifier__learning_rate': loguniform(0.01, 0.3),
    'classifier__max_depth': randint(3, 10),
    'classifier__min_child_weight': randint(1, 10),
    'classifier__subsample': uniform(0.6, 0.4),          # 0.6 - 1.0
    'classifier__colsample_bytree': uniform(0.6, 0.4),   # 0.6 - 1.0
    'classifier__gamma': uniform(0, 5),
    'classifier__reg_alpha': loguniform(1e-3, 10),
    'classifier__reg_lambda': loguniform(1e-2, 10),
}

cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

XGBoost_search = RandomizedSearchCV(
    pipeline,
    param_distributions=param_distributions,
    n_iter=30,
    cv=cv,
    scoring='accuracy',
    n_jobs=-1,          # parallelni CV darajasida qilamiz, XGBoost ichida n_jobs=1
    random_state=42,
    verbose=2,
    refit=True,
)

XGBoost_search.fit(X_train, Y_train_enc)

print("Best parameters:", XGBoost_search.best_params_)
print("Best CV accuracy:", round(XGBoost_search.best_score_, 4))
```

```python
Best_params = {
    'SVC': svc_grid_search.best_params_,
    'KNN': knn_grid_search.best_params_,
    'CatBoost': Catboost_search.best_params_,
    'XGBoost': XGBoost_search.best_params_,
}
```

```python
import numpy as np
from sklearn.base import clone
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from catboost import CatBoostClassifier
from xgboost import XGBClassifier

def clean(p):
    out = {}
    for k, v in p.items():
        k = k.replace('classifier__', '')
        if isinstance(v, np.generic):
            v = v.item()          # np.float64 -> float, np.int64 -> int
        out[k] = v
    return out

models = {
    'SVC': SVC(**clean(Best_params_dict['SVC']), probability=True, random_state=42),
    'KNN': KNeighborsClassifier(**clean(Best_params_dict['KNN'])),
    'CatBoost': CatBoostClassifier(**clean(Best_params_dict['CatBoost']),
                                   verbose=0, random_state=42, thread_count=1),
    'XGBoost': XGBClassifier(**clean(Best_params_dict['XGBoost']),
                             eval_metric='logloss', random_state=42, n_jobs=1),
}

pipelines = {
    name: Pipeline([('preprocessor', clone(Transformer)), ('classifier', m)])
    for name, m in models.items()
}

# Tekshiruv: clone xatosiz o'tishi va np.float64 yo'qligi kerak
print(pipelines['CatBoost'].named_steps['classifier'].get_params())
clone(pipelines['CatBoost'])
print("clone OK")
```

```python

from sklearn.model_selection import cross_validate


rows, cv_acc = [], {}
for name, pipe in pipelines.items():
    res = cross_validate(pipe, X_train, Y_train_enc, cv=cv, scoring=scoring, n_jobs=-1)
    cv_acc[name] = res['test_accuracy']
    rows.append({
        'Model': name,
        'Accuracy': res['test_accuracy'].mean(),
        'Acc std': res['test_accuracy'].std(),
        'F1': res['test_f1'].mean(),
        'ROC AUC': res['test_roc_auc'].mean(),
        'Fit time (s)': res['fit_time'].mean(),
    })
```

```python

comparison = pd.DataFrame(rows).set_index('Model').sort_values('ROC AUC', ascending=False)
print(comparison.round(4))

# 5. Eng kuchlisini tanlash (ROC AUC bo'yicha; farq std dan kichik bo'lsa, tezroq modelni oling)
best_name = comparison.index[0]
top = comparison.iloc[0]
close = comparison[(top['ROC AUC'] - comparison['ROC AUC']) < comparison['Acc std']]
if len(close) > 1:
    best_name = close['Fit time (s)'].idxmin()
    print(f"\nNatijalar yaqin, eng tez model tanlandi: {best_name}")
print(f"Tanlangan model: {best_name}")
```

```python
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, classification_report

final_model = pipelines['SVC'].fit(X_train, Y_train_enc)
pred = final_model.predict(X_test)
proba = final_model.predict_proba(X_test)[:, 1]

print("\nTest accuracy:", round(accuracy_score(Y_test_enc, pred), 4))
print("Test F1:", round(f1_score(Y_test_enc, pred), 4))
print("Test ROC AUC:", round(roc_auc_score(Y_test_enc, proba), 4))
print(classification_report(Y_test_enc, pred, target_names=le.classes_.astype(str)))
```
