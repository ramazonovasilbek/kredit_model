```python


import pickle


with open("final_model.pkl", "rb") as f:
    model = pickle.load(f)

with open("LabelEncoder.pkl", "rb") as f:
    le = pickle.load(f)
```

```python
import numpy as np
import pandas as pd



def predict(data):
    if isinstance(data, dict):
        df = pd.DataFrame([data])
    elif isinstance(data, list):
        df = pd.DataFrame(data)
    else:
        df = data.copy()

    pred = np.ravel(model.predict(df)).astype(int)       # 0/1
    labels = le.inverse_transform(pred)                  # 'N'/'Y'

    result = pd.DataFrame({'prediction': labels})

    # ehtimollik (agar model qo'llab-quvvatlasa)
    if hasattr(model, "predict_proba"):
        proba = model.predict_proba(df)
        for i, cls in enumerate(le.classes_):
            result[f'proba_{cls}'] = proba[:, i].round(4)

    return result
```

```python
data = {
    'Jinsi': 'Male',
    "Uylangan": 'No',
    "Qarindoshlari soni": '0',
    "Ma'lumoti": 'Graduate',
    "O'zini o'zi ish bilan ta'minlagan": "Yes",
    "Ariza beruvchining daromadi": 10000,
    "Hamkorlikdagi ariza beruvchining daromadi": 10000,
    "Kredit miqdori": 5000,
    "Kredit miqdorining muddati": 40,
    "Kredit Tarixi": 1,
    "Mulk Maydoni": "Urban"
}

print(predict(data))
```
